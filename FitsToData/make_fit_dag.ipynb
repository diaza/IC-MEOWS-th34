```python
import numpy as np
import utils
import make_dag
```

```python
logdm2_list = np.asarray(np.ndarray.tolist(np.arange(-2.10, 2.10, 0.1))+ [2.15])
dm2_list = np.power(10, logdm2_list)  
dm2_list = dm2_list[np.logical_and(dm2_list>.1, dm2_list<50)]
dm2_list = np.append(dm2_list, 50.0)

log_Um4sq_list = np.asarray(np.ndarray.tolist(np.arange(-3., 0., 0.1)))  
Um4sq_list = np.power(10, log_Um4sq_list)
Um4sq_list = Um4sq_list[Um4sq_list<0.50]
Um4sq_list = np.append(Um4sq_list, 0.50)

log_Ut4sq_list = np.asarray(np.ndarray.tolist(np.arange(-3., 0., 0.2)))   
Ut4sq_list = np.power(10, log_Ut4sq_list)  
Ut4sq_list = np.insert(Ut4sq_list, 0, 0.)
Ut4sq_list = Ut4sq_list[Ut4sq_list<0.50]
Ut4sq_list = np.append(Ut4sq_list, 0.50)

d24_list = [np.pi]
```

```python
realization_file = "/data/ana/SterileNeutrino/IC86/HighEnergy/Analysis/scripts/jobs/Data/Data_IC86.All_cut_everything_platinum_LT_240747841.npz"
def make_fit_dag(n_seeds = "-1", FS = '0.50', cr_width = "0.03"):
    
    dag_filename ='./FS_{FS}/n_seeds_{n_seeds}/fittodata_n_seeds_{n_seeds}_crwidth_{cr_width}.dag'.format(FS=FS,n_seeds=n_seeds, cr_width = cr_width)
    
    dirname = "/data/ana/SterileNeutrino/IC86/HighEnergy/Analysis/th34/Fits/FitToData/AIRS_mceq121_2011_sib_HG_E3/crSlopeWidth_{}/FS_{}/n_seeds_{}/".format(cr_width, FS,n_seeds)

    m = make_dag.DagMaker(fast_scaling = FS, realization_file = realization_file,
                                  osg = 'True', do_fit = True, n_seeds = n_seeds, crSlopeWidth = cr_width)
    
    Ue4sq= 0.
    del14=0.
    with open(dag_filename, 'w') as f:
        i=0
        for Ut4sq in Ut4sq_list:
            
            if Ut4sq == 0.:
                m.addtodag(dm2=0., Ue4sq=0., Um4sq=0., Ut4sq=0., del14=0., del24=0., f=f, 
                           outdir = dirname + "/Ut4sq_{:.6f}/".format(Ut4sq), job_index = i)
                i+=1

                
            for Um4sq in Um4sq_list:
                for dm2 in dm2_list:
                    for del24 in d24_list:

                        m.addtodag(dm2=dm2, Ue4sq=Ue4sq, Um4sq=Um4sq, Ut4sq=Ut4sq, del14=del14,        
                                   del24=del24, f=f, 
                                   outdir = dirname + "/Ut4sq_{:.6f}/".format(Ut4sq),
                                   job_index = i)
                        i+=1
```

```python
make_fit_dag(n_seeds = "-1", FS = '0.50', cr_width = "0.25")
make_fit_dag(n_seeds = "4", FS = '0.50', cr_width = "0.25")
```
