```python
fig, ax = plt.subplots(figsize = (12,6))

sys_names = list(utils.sys_dict.keys())
sys_strings = [utils.prop_translator[sys_name] for sys_name in sys_names]
sys_values = np.array([BF_data[k] for k in sys_names])
sys_CV = [BF_data[k] for k in sys_names]
sys_diff = sys_values - [utils.sys_dict[k][0] for k in sys_names]
pull = sys_diff / [utils.sys_dict[k][1] for k in sys_names]

ax.tick_params(axis='x', labelrotation=90)
upper_bound = (np.array([utils.sys_dict[k][3] for k in sys_names]) - [utils.sys_dict[k][0] for k in sys_names])/[utils.sys_dict[k][1] for k in sys_names]
lower_bound = (np.array([utils.sys_dict[k][2] for k in sys_names]) - [utils.sys_dict[k][0] for k in sys_names])/[utils.sys_dict[k][1] for k in sys_names]
upper_bound[np.isinf(upper_bound)]=100

ax.bar(sys_strings, pull, label = r"Best Fit")
p = ax.bar(sys_strings, upper_bound, width = 0.5, alpha = 0.3, label="Bounds", zorder = 0)
ax.bar(sys_strings, lower_bound, color = p[0].get_facecolor(), width = 0.5, alpha = 0.3, zorder = 0)

sys_values_th34_0 = np.array([th34_0_min_data[k] for k in sys_names])
sys_CV_th34_0 = [th34_0_min_data[k] for k in sys_names]
sys_diff_th34_0 = sys_values_th34_0 - [utils.sys_dict[k][0] for k in sys_names]
pull_th34_0 = sys_diff_th34_0 / [utils.sys_dict[k][1] for k in sys_names]


ax.plot(sys_strings, pull_th34_0, 'k_', markersize = 20, markeredgewidth = 2, label=r"$\theta_{34}=0$")
# ax.plot(sys_strings, lower_bound, 'k_', markersize = 20, markeredgewidth = 2)

ax.set_title("Systematic Pulls")
ax.minorticks_off()
ax.set_ylim(-3,3.2)
ax.set_ylabel("Pulls [$\sigma$]")
ax.grid(alpha = .5)
ax.legend(loc = "upper right", frameon = True)
plt.tight_layout()
# plt.savefig("syspulls.png", dpi = 144)
plt.show()
```
